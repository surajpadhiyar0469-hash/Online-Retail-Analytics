```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel("../dataset/online_retail.xlsx")
```

```python
# 1.Sabse zyada revenue generate karne wala country find karo.
df['revenue']=df['Quantity']*df['Price']
country=df.groupby('Country')['revenue'].sum()
top_country=country.sort_values(ascending=False).head(1)
print(top_country)
```

```python
# 2.Sabse zyada revenue generate karne wala product find karo.
product=df.groupby('Description')['revenue'].sum()
top_product=product.sort_values(ascending=False).head(1)
print(top_product)
```

```python
# 3.Sabse valuable customer find karo.
customer=df.groupby('Customer ID')['revenue'].sum()
top_customer=customer.sort_values(ascending=False).head(1)
print(top_customer)
```

```python
# 4.Highest sales wala month find karo.
df['month']=df['InvoiceDate'].dt.month
month=df.groupby('month')['revenue'].sum()
top_month=month.sort_values(ascending=False).head(1)
print(top_month)
```

```python
# 5.Highest sales wala year find karo.
df['year']=df['InvoiceDate'].dt.year
year=df.groupby('year')['revenue'].sum()
top_year=year.sort_values(ascending=False).head(1)
print(top_year)
```

```python
# 6.Aise customers find karo jinhone ek se zyada invoices generate kiye.
invoice_customer=df.groupby('Customer ID')['Invoice'].nunique()
print(invoice_customer[invoice_customer>1])
```

```python
# 7.Aise products find karo jinki total Quantity negative hai.
negative_quantity=df.groupby('Description')['Quantity'].sum()
print(negative_quantity[negative_quantity<0])
```

```python
# 8.Country-wise average invoice value calculate karo.
ave_country_invoice=df.groupby('Country')['revenue'].mean()
print(ave_country_invoice)
```

```python
# 9.Top 10 customers ka total revenue aur percentage contribution calculate karo.
top10_customer=df.groupby('Customer ID')['revenue'].sum()
top10=top10_customer.sort_values(ascending=False).head(10)
print("In value:")
print(top10)
total_revenue=df['revenue'].sum()
print("In Percentage:")
print((top10/total_revenue)*100)
```
