```python
# Q25. Data Filtering

# 1. Aise records find karo jinki Quantity 10 se greater hai.
df[df['Quantity']>10]
```

```python
# 2. Aise records find karo jinki UnitPrice 100 se greater hai.
df[df['Price']>100]
```

```python
#  3. Aise records find karo jinki Quantity 10 se greater aur UnitPrice 100 se greater hai.
df[(df['Quantity']>10)&(df['Price']>100)]
```

```python
# 4. Aise records find karo jinki Country "United Kingdom" nahi hai.
df[df['Country']!="United Kingdom"]
```

```python
# Q27. GroupBy Analysis

# 1. Country-wise total Quantity calculate karo.
df.groupby("Country")["Quantity"].sum()
```

```python
# 2. Country-wise average UnitPrice calculate karo.
df.groupby("Country")["Price"].mean()
```

```python
# 3. Top 10 countries find karo jahan sabse zyada Quantity sold hui.
a=df.groupby("Country")["Quantity"].sum()
top10=a.sort_values(ascending=False).head(10)
top10
```

```python
# 4. Country-wise total Revenue calculate karo.
df['revenue']=df['Quantity']*df['Price']
df.groupby('Country')['revenue'].sum()
```

```python

# 5. Top 10 countries find karo jahan sabse zyada Revenue generate hua.
a=df.groupby('Country')['revenue'].sum()
top10=a.sort_values(ascending=False).head(10)
top10
```
