```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel("../dataset/online_retail.xlsx")
```

```python
# Assignment 1 – Basic Visualizations
# - Q1. Monthly Revenue ka Line Chart banao.
df['revenue']=df['Price']*df['Quantity']
df['month']=df['InvoiceDate'].dt.month
monthly_revenue=df.groupby('month')['revenue'].sum()
plt.plot(monthly_revenue)
```

```python
# - Q2. Top 10 Products ka Bar Chart banao.
products=df.groupby('Description')['Quantity'].sum()
top10_products=products.sort_values(ascending=False).head(10)
print(top10_products)
plt.bar(top10_products.index,top10_products.values)
```

```python
# - Q3. Top 10 Countries by Revenue ka Bar Chart banao.
country=df.groupby('Country')['revenue'].sum()
top10_country=country.sort_values(ascending=False).head(10)
print(top10_country)
plt.bar(top10_country.index,top10_country.values)
```

```python
# - Q4. Revenue ka Histogram banao.
plt.hist(df['revenue'])
```

```python
# - Q5. Quantity ka Box Plot banao.
plt.boxplot(df['Quantity'])
```

```python
# Assignment 2 – Intermediate Visualizations
# Q6. Year-wise Revenue ka Bar Chart banao.
df['year']=df['InvoiceDate'].dt.year
year_revenue=df.groupby('year')['revenue'].sum()
plt.bar(year_revenue.index,year_revenue.values)
```

```python
# Q7. Top 10 Customers by Revenue ka Horizontal Bar Chart banao.
customer=df.groupby('Customer ID')['revenue'].sum()
top10_customer=customer.sort_values(ascending=False).head(10)
print(top10_customer)
plt.barh(top10_customer.index.astype(str),top10_customer.values)
```

```python
# Q8. Country-wise Revenue ka Pie Chart banao (Top 5 Countries).
top5_country=country.sort_values(ascending=False).head(5)
plt.pie(top5_country,labels=top5_country.index,autopct="%1.1f%%")
```

```python
# Q9. Quantity aur Price ka Scatter Plot banao.
plt.scatter(x=df['Quantity'],y=df['Price'])
```

```python
# Q10. Monthly Revenue ka Seaborn Bar Plot banao.
sns.barplot(monthly_revenue)
```

```python
# Assignment 3 – Advanced Visualizations
# Q11. Monthly Revenue ka Seaborn Line Plot banao.
sns.lineplot(monthly_revenue)
```

```python
# Q12. Revenue aur Quantity ka Correlation Heatmap banao.
graph=df[['Quantity','Price']].corr()
sns.heatmap(graph,annot=True)
```

```python
# Q13. Top 10 Products ka Seaborn Count Plot banao.
sns.countplot(top10_products.values)
# Q15. Monthly Revenue aur Quantity ka Dual Visualization banao.
```

```python
# Q14. Country-wise Revenue ka Violin Plot banao (Top 5 Countries).
sns.violinplot(top5_country)
```
