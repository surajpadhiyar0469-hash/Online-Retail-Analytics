```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel("../dataset/online_retail.xlsx")
```

```python
# 1 – Data Cleaning Practice
# 1. Dataset mein missing values identify karo.
df.isnull()

```

```python
# 2. Har column mein missing values ki count find karo.
df.isnull().sum()
```

```python
# 3. Missing values ka percentage calculate karo.
df.isnull().mean()
```

```python
# 4. Duplicate records identify karo.
df.duplicated()
```

```python
# 5. Duplicate records remove karo.
rf1=df.copy()
rf1=rf1.drop_duplicates()
rf1
```

```python
# 6. Description column ke missing values handle karo.
rf1 = df.copy()

rf1['Description'] = rf1['Description'].fillna(
    rf1['Description'].mode()[0]
)

print(rf1.isnull().sum())

```

```python
import pandas as pd

print(pd.__version__)
print(type(rf1))
print(rf1.__copy__)
```

```python
# 7.Customer ID column ke missing values handle karo.
rf1['Customer ID']=rf1['Customer ID'].fillna(rf1['Customer ID'].mean())
print(rf1.isnull().sum())
```

```python
# 8.InvoiceDate ka datatype check karo.
print(df['InvoiceDate'].dtype)
```

```python
df.dtypes
```

```python
# 9. InvoiceDate ko datetime mein convert karo.
rf1['InvoiceDate']=pd.to_datetime(rf1['InvoiceDate'])
rf1.dtypes
```

```python
# 10. Negative Quantity aur Price wale records identify karo.
print("negative Quantity:",(df['Quantity']<0).sum())
print("negative Price:",(df['Price']<0).sum())
```

```python
# Assignment 2 – Advanced Data Cleaning 

# 1. Description column ke extra spaces remove karo.
rf1['Description']=rf1['Description'].str.strip()
print(rf1['Description'])
```

```python
# 2. Description column ke sabhi text ko uppercase mein convert karo.
rf1['Description']=rf1['Description'].str.upper()
print(rf1['Description'])
```

```python
# 3. InvoiceDate column se Year extract karo.
rf1['year']=rf1['InvoiceDate'].dt.year
print(rf1['year'])
```

```python
# 4. InvoiceDate column se Month extract karo.
rf1['month']=rf1['InvoiceDate'].dt.month
print(rf1['month'])
```

```python
# 5. InvoiceDate column se Day extract karo.
rf1['day']=rf1['InvoiceDate'].dt.day
print(rf1['day'])
```

```python
# 6. Quantity column mein zero values wale records identify karo.
print("Quntity Column Zero Values sum:",(df['Quantity']==0).sum())
```

```python
# 7. Price column mein zero values wale records identify karo.
print("Price Column Total Zero Values:",(df['Price']==0).sum())
```

```python
# 8. Negative Revenue wale records identify karo.
rf1['revenue']=df['Price']*df['Quantity']

print("Total Negative Revenue:",(rf1['revenue']<0).sum())
```

```python
# 9. StockCode column mein unique values ki total count find karo.
print(df['StockCode'].nunique())
```
