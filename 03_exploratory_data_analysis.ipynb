```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel("../dataset/online_retail.xlsx")
```

```python
# Q1. Sales Analysis
# 1. Total Revenue calculate karo.
df['revenue']=df['Price']*df['Quantity']
print("Total Revenue:",df['revenue'].sum())
```

```python
# 2. Total Quantity sold calculate karo.
print("Total Quantity:",df['Quantity'].sum())
```

```python
# 3. Total invoices count find karo.
print("Total Invoice",df['Invoice'].nunique())
```

```python
# Q2. Product Analysis

# 1. Top 10 most sold products find karo.
product=df.groupby('Description')['Quantity'].sum()
top10_products=product.sort_values(ascending=False).head(10)
print(top10_products)
```

```python
# 2. Top 10 products by Revenue find karo.
revenue_product=df.groupby('Description')['revenue'].sum()
top10=revenue_product.sort_values(ascending=False).head(10)
print(top10)
```

```python
# Q3. Country Analysis
# 1. Country-wise total Revenue calculate karo.
country_revenue1=df.groupby('Country')['revenue'].sum()
country=country_revenue1.sort_values()
print(country)
```

```python
# 2. Top 10 countries by Revenue find karo.
country_revenue=df.groupby('Country')['revenue'].sum()
top10_country=country_revenue.sort_values(ascending=False).head(10)
print(top10_country)
```

```python
# Q4. Customer Analysis
# 1. Total unique customers find karo.
print(df['Customer ID'].nunique())
```

```python
# 2. Top 10 customers by Revenue find karo.
customer=df.groupby('Customer ID')['revenue'].sum()
top10_customer=customer.sort_values(ascending=False).head(10)
print(top10_customer)
```

```python
# Q5. Monthly Sales Analysis
# 1. Month-wise Revenue calculate karo.
df['month']=df['InvoiceDate'].dt.month
month=df.groupby('month')['revenue'].sum()
month_sort=month.sort_values(ascending=True)
print(month_sort)
```

```python
# 2. Highest Revenue wala month find karo.
print(month.idxmax())
```

```python
# Q6. Yearly Sales Analysis
# 1. Year-wise Revenue calculate karo.
df['year']=df['InvoiceDate'].dt.year
year_revenue=df.groupby('year')['revenue'].sum()
print(year_revenue)
```

```python
# 2. Har year ki total Quantity find karo.
year_quantity=df.groupby('year')['Quantity'].sum()
print(year_quantity)
```

```python
# Q7. Invoice Analysis
# 1. Average Invoice Value calculate karo.
invoice_revenue=df.groupby('Invoice')['revenue'].sum()
print(invoice_revenue.mean())
```

```python
# 2. Highest Value wala Invoice find karo.
print(invoice_revenue.idxmax())
print("revenue:",invoice_revenue.max())
```

```python
# Q8. Price Analysis
# 1. Average Price calculate karo.
print(df['Price'].mean())
```

```python
# 2. Highest aur Lowest Price find karo.
print('Highest Price:',df['Price'].max())
print("lowest Price:",df['Price'].min())
```

```python
# Q9. Quantity Analysis
# 1. Average Quantity calculate karo.
print('Average Of Quantity:',df['Quantity'].mean())
```

```python
# 2. Highest Quantity wale 10 records find karo.
quantity=df.sort_values('Quantity',ascending=False).head(10)
print(quantity)
```

```python

# Q10. Category and Country Insights
# 1. Country-wise unique customers count karo.
c_customer=df.groupby('Country')['Customer ID'].nunique()
print(c_customer)
```

```python
# 2. Country-wise unique products count karo.
c_products=df.groupby('Country')['Description'].nunique()
print(c_products)
```
